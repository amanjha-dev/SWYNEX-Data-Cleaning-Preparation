```python
import pandas as pd
import numpy as np

# Load the original dataset
df = pd.read_csv("../data/raw_dataset.csv")

# Display the first five rows
display(df.head())
```

```python
# Number of rows and columns
print("Dataset shape:", df.shape)

# Column names
print("\nColumn names:")
print(df.columns.tolist())

# Data types and non-null counts
print("\nDataset information:")
df.info()

# Missing values in each column
print("\nMissing values:")
print(df.isnull().sum())

# Duplicate rows
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df.shape

df.isnull().sum()

df.duplicated().sum()

df.info()
```

```python
# Count completely empty rows
empty_rows = df.isnull().all(axis=1).sum()

print("Completely empty rows:", empty_rows)

# Count fully duplicated rows
duplicate_rows = df.duplicated().sum()

print("Fully duplicated rows:", duplicate_rows)

# Inspect rows with missing order IDs
print("\nRows with missing order IDs:")
display(df[df["order_id"].isnull()].head(10))
```

```python
# Store the original number of rows
original_rows = len(df)

# Remove completely empty rows
df = df.dropna(how="all")

# Count rows after removing empty rows
rows_after_empty_removal = len(df)

# Remove exact duplicate records
df = df.drop_duplicates()

# Count rows after removing duplicates
cleaned_rows = len(df)

# Display the results
print("Original rows:", original_rows)
print("Completely empty rows removed:",
      original_rows - rows_after_empty_removal)
print("Duplicate rows removed:",
      rows_after_empty_removal - cleaned_rows)
print("Rows remaining:", cleaned_rows)
```

```python

# Inspect categorical columns for inconsistent values

categorical_columns = [
    "gender",
    "region",
    "city",
    "product_category",
    "product_name",
    "payment_method",
    "order_status",
    "return_flag"
]

for col in categorical_columns:
    print(f"\n--- {col} ---")
    print(df[col].value_counts(dropna=False).head(25))
```

```python

# Remove leading/trailing spaces from text columns
for col in categorical_columns:
    df[col] = df[col].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

# Standardize gender values
gender_mapping = {
    "male": "Male",
    "m": "Male",
    "female": "Female",
    "f": "Female",
    "other": "Other"
}

df["gender"] = df["gender"].apply(
    lambda x: gender_mapping.get(x.lower(), x)
    if isinstance(x, str) else x
)

# Check the cleaned gender values
print(df["gender"].value_counts(dropna=False))
```

```python

# Inspect different date formats
print("Sample dates:")
print(df["order_date"].head(15).to_list())

print("\nData type before cleaning:")
print(df["order_date"].dtype)

print("\nMissing dates:")
print(df["order_date"].isna().sum())

print("\nExample distinct dates:")
print(df["order_date"].dropna().drop_duplicates().head(15).to_list())
```

```python

# Remove any accidental spaces
date_text = df["order_date"].astype("string").str.strip()

# Identify dates in YYYY-MM-DD format
iso_format = date_text.str.match(
    r"^\d{4}-\d{2}-\d{2}$", na=False
)

# Create a datetime column
clean_dates = pd.Series(
    pd.NaT, index=df.index, dtype="datetime64[ns]"
)

# Parse YYYY-MM-DD dates
clean_dates.loc[iso_format] = pd.to_datetime(
    date_text.loc[iso_format],
    format="%Y-%m-%d",
    errors="coerce"
)

# Parse DD/MM/YYYY dates
clean_dates.loc[~iso_format] = pd.to_datetime(
    date_text.loc[~iso_format],
    format="%d/%m/%Y",
    errors="coerce"
)

# Update the dataset
df["order_date"] = clean_dates

# Verify the conversion
print("Data type after cleaning:", df["order_date"].dtype)
print("Missing dates after conversion:", df["order_date"].isna().sum())
print("\nSample cleaned dates:")
print(df["order_date"].head(15))
```

```python

# Reload only the original date column
raw_dates = pd.read_csv(
    "../data/raw_dataset.csv",
    usecols=["order_date"]
)["order_date"].astype("string").str.strip()

# Identify the two expected formats
iso_format = raw_dates.str.match(
    r"^\d{4}-\d{2}-\d{2}$", na=False
)

dmy_format = raw_dates.str.match(
    r"^\d{2}/\d{2}/\d{4}$", na=False
)

# Display format counts
print("YYYY-MM-DD format:", iso_format.sum())
print("DD/MM/YYYY format:", dmy_format.sum())
print("Missing or unexpected formats:",
      (~(iso_format | dmy_format)).sum())

# Show unexpected values
unexpected_dates = raw_dates[~(iso_format | dmy_format)]

print("\nUnexpected date values:")
print(unexpected_dates.value_counts(dropna=False).head(30))
```

```python

# Create an empty datetime series
clean_dates = pd.Series(
    pd.NaT, index=raw_dates.index, dtype="datetime64[ns]"
)

# Format 1: YYYY-MM-DD
iso_format = raw_dates.str.match(
    r"^\d{4}-\d{2}-\d{2}$", na=False
)
clean_dates.loc[iso_format] = pd.to_datetime(
    raw_dates.loc[iso_format],
    format="%Y-%m-%d",
    errors="coerce"
)

# Format 2: DD/MM/YYYY
dmy_format = raw_dates.str.match(
    r"^\d{2}/\d{2}/\d{4}$", na=False
)
clean_dates.loc[dmy_format] = pd.to_datetime(
    raw_dates.loc[dmy_format],
    format="%d/%m/%Y",
    errors="coerce"
)

# Format 3: Month DD, YYYY
written_format = ~(iso_format | dmy_format) & raw_dates.notna()
clean_dates.loc[written_format] = pd.to_datetime(
    raw_dates.loc[written_format],
    format="%B %d, %Y",
    errors="coerce"
)

# Update the cleaned dataset
df["order_date"] = clean_dates

# Verify results
print("Original missing dates:", raw_dates.isna().sum())
print("Missing dates after cleaning:", df["order_date"].isna().sum())
print("Valid dates:", df["order_date"].notna().sum())
print("Final date data type:", df["order_date"].dtype)

# Check whether any non-missing source dates failed to parse
failed_dates = raw_dates.notna() & df["order_date"].isna()
print("\nNon-missing dates that failed to parse:", failed_dates.sum())
print(raw_dates[failed_dates].value_counts().head(20))
```

```python

print("Current dataset rows:", len(df))
print("Current missing dates:", df["order_date"].isna().sum())

# Compare the original row indices still present in df
print("Current index range:", df.index.min(), "to", df.index.max())
print("Missing dates in original source:", raw_dates.isna().sum())
```

```python

# Inspect age statistics
print("Age data type:", df["age"].dtype)
print("Missing ages:", df["age"].isna().sum())

print("\nAge statistics:")
print(df["age"].describe())

print("\nLowest ages:")
print(df["age"].value_counts().sort_index().head(20))

print("\nHighest ages:")
print(df["age"].value_counts().sort_index().tail(20))

print("\nAges below 18:")
print((df["age"] < 18).sum())

print("\nAges above 100:")
print((df["age"] > 100).sum())
```

```python

# Count invalid ages before cleaning
invalid_age_mask = (df["age"] < 18) | (df["age"] > 100)

print("Invalid ages before cleaning:", invalid_age_mask.sum())
print("\nExamples of invalid ages:")
print(df.loc[invalid_age_mask, "age"].value_counts().sort_index())

# Replace invalid ages with missing values
df.loc[invalid_age_mask, "age"] = np.nan

# Calculate median using valid ages only
median_age = df["age"].median()

print("\nMedian age of valid records:", median_age)
print("Missing ages after validation:", df["age"].isna().sum())

```

```python
# Fill missing ages using the median
df["age"] = df["age"].fillna(median_age)

# Verify the result
print("Missing ages after imputation:", df["age"].isna().sum())
print("Median used for imputation:", median_age)
print("\nUpdated age statistics:")
print(df["age"].describe())
```

```python
# Inspect quantity column
print("Data type:", df["quantity"].dtype)
print("Missing quantities:", df["quantity"].isna().sum())

print("\nQuantity statistics:")
print(df["quantity"].describe())

print("\nQuantity values in ascending order:")
print(df["quantity"].value_counts().sort_index().head(20))

print("\nZero or negative quantities:")
print((df["quantity"] <= 0).sum())
```

```python
# Count suspicious quantities before cleaning
print("Negative or zero quantities:", (df["quantity"] <= 0).sum())
print("Quantities equal to 999:", (df["quantity"] == 999).sum())

# Replace invalid quantities with missing values
invalid_quantity = (df["quantity"] <= 0) | (df["quantity"] == 999)
df.loc[invalid_quantity, "quantity"] = np.nan

# Calculate median from valid quantities
median_quantity = df["quantity"].median()

print("\nMedian valid quantity:", median_quantity)
print("Missing quantities after validation:", df["quantity"].isna().sum())
```

```python
# Fill missing quantities using the median
df["quantity"] = df["quantity"].fillna(median_quantity)

# Verify the result
print("Missing quantities after imputation:", df["quantity"].isna().sum())
print("Median used:", median_quantity)

print("\nUpdated quantity statistics:")
print(df["quantity"].describe())

print("\nUnique quantity values:")
print(sorted(df["quantity"].unique()))
```

```python
# Inspect discount_pct
print("Data type:", df["discount_pct"].dtype)
print("Missing discounts:", df["discount_pct"].isna().sum())

print("\nDiscount statistics:")
print(df["discount_pct"].describe())

print("\nLowest discount values:")
print(df["discount_pct"].value_counts().sort_index().head(15))

print("\nDiscounts below 0:", (df["discount_pct"] < 0).sum())
print("Discounts above 100:", (df["discount_pct"] > 100).sum())
```

```python
# Calculate the median from existing valid discounts
median_discount = df["discount_pct"].median()

print("Median discount:", median_discount)

# Fill missing discounts using the median
df["discount_pct"] = df["discount_pct"].fillna(median_discount)

# Verify the result
print("Missing discounts after imputation:", df["discount_pct"].isna().sum())
print("\nUpdated discount statistics:")
print(df["discount_pct"].describe())
```

```python
# Inspect customer satisfaction
print("Data type:", df["customer_satisfaction"].dtype)
print("Missing ratings:", df["customer_satisfaction"].isna().sum())

print("\nRating statistics:")
print(df["customer_satisfaction"].describe())

print("\nRating value counts:")
print(df["customer_satisfaction"].value_counts().sort_index())

print("\nRatings below 1:", (df["customer_satisfaction"] < 1).sum())
print("Ratings above 5:", (df["customer_satisfaction"] > 5).sum())
```

```python
# Keep missing ratings as missing; verify the valid range
valid_ratings = df["customer_satisfaction"].dropna()

print("Missing ratings retained:", df["customer_satisfaction"].isna().sum())
print("All observed ratings between 1 and 5:",
      valid_ratings.between(1, 5).all())
print("Number of valid ratings:", valid_ratings.count())
```

```python
# Inspect shipping time
print("Data type:", df["days_to_ship"].dtype)
print("Missing shipping times:", df["days_to_ship"].isna().sum())

print("\nShipping-time statistics:")
print(df["days_to_ship"].describe())

print("\nShipping-time value counts:")
print(df["days_to_ship"].value_counts().sort_index())

print("\nZero or negative shipping times:",
      (df["days_to_ship"] <= 0).sum())

print("\nShipping time by order status:")
print(
    df.groupby("order_status", dropna=False)["days_to_ship"]
      .agg(["count", "mean", "min", "max"])
)
```

```python
# Count unusual shipping times
print("Negative shipping times:", (df["days_to_ship"] < 0).sum())
print("Zero shipping times:", (df["days_to_ship"] == 0).sum())
print("Shipping times above 30 days:", (df["days_to_ship"] > 30).sum())

# Display records with unusually high shipping times
print("\nRecords with shipping times above 30 days:")
print(
    df.loc[
        df["days_to_ship"] > 30,
        ["order_id", "order_status", "days_to_ship"]
    ]
)
```

```python
# Standardize order status capitalization
df["order_status"] = df["order_status"].apply(
    lambda x: x.strip().capitalize()
    if isinstance(x, str) else x
)

# Count shipping times outside the chosen plausible range
invalid_shipping = (
    (df["days_to_ship"] < 1) |
    (df["days_to_ship"] > 30)
)

print("Invalid shipping times:", invalid_shipping.sum())

# Replace invalid shipping times with missing values
df.loc[invalid_shipping, "days_to_ship"] = np.nan

# Check results
print("Missing shipping times after validation:",
      df["days_to_ship"].isna().sum())

print("\nStandardized order statuses:")
print(df["order_status"].value_counts(dropna=False))

print("\nShipping-time statistics after validation:")
print(df["days_to_ship"].describe())
```

```python
# Examine missing shipping times by order status
print("Missing shipping times by order status:")
print(
    df.groupby("order_status")["days_to_ship"]
      .agg(
          total_orders="size",
          missing_shipping_times=lambda x: x.isna().sum()
      )
)

print("\nMissing shipping time records:")
print(
    df.loc[
        df["days_to_ship"].isna(),
        ["order_id", "order_status", "days_to_ship"]
    ]["order_status"].value_counts()
)
```

```python
# Keep unknown shipping times as missing
print("Total orders:", len(df))
print("Missing shipping times retained:", df["days_to_ship"].isna().sum())
print("Valid shipping times:", df["days_to_ship"].notna().sum())

# Confirm remaining shipping times are within the selected range
valid_shipping = df["days_to_ship"].dropna()

print(
    "All valid shipping times between 1 and 30 days:",
    valid_shipping.between(1, 30).all()
)
```

```python
financial_columns = [
    "unit_price",
    "sales_amount",
    "profit",
    "shipping_cost"
]

print("Missing values:")
print(df[financial_columns].isna().sum())

print("\nFinancial statistics:")
print(df[financial_columns].describe())

print("\nNumber of negative values:")
print((df[financial_columns] < 0).sum())

print("\nNumber of zero values:")
print((df[financial_columns] == 0).sum())
```

```python
# Inspect negative shipping costs
print("Records with negative shipping costs:")
print(
    df.loc[
        df["shipping_cost"] < 0,
        ["order_id", "quantity", "unit_price",
         "discount_pct", "sales_amount", "profit", "shipping_cost"]
    ]
)

# Inspect the five largest sales amounts
print("\nFive largest sales amounts:")
print(
    df.nlargest(5, "sales_amount")[
        ["order_id", "quantity", "unit_price",
         "discount_pct", "sales_amount", "profit", "shipping_cost"]
    ]
)
```

```python
# Calculate expected sales using quantity, unit price, and discount
df["expected_sales"] = (
    df["quantity"]
    * df["unit_price"]
    * (1 - df["discount_pct"])
)

# Compare expected and recorded sales
df["sales_difference"] = (
    df["sales_amount"] - df["expected_sales"]
)

print("Five largest recorded sales amounts:")
print(
    df.nlargest(5, "sales_amount")[
        ["order_id", "quantity", "unit_price", "discount_pct",
         "expected_sales", "sales_amount", "sales_difference"]
    ].round(2)
)

print("\nRecords where recorded sales differ from expected sales by over 10%:")
difference_pct = (
    df["sales_difference"].abs()
    / df["expected_sales"].abs()
)
print((difference_pct > 0.10).sum())
```

```python
# Count negative shipping costs before cleaning
negative_shipping_costs = (df["shipping_cost"] < 0).sum()

print("Negative shipping costs before cleaning:",
      negative_shipping_costs)

# Replace negative shipping costs with missing values
df.loc[df["shipping_cost"] < 0, "shipping_cost"] = np.nan

# Verify
print("Negative shipping costs after cleaning:",
      (df["shipping_cost"] < 0).sum())

print("Missing shipping costs after cleaning:",
      df["shipping_cost"].isna().sum())
```

```python
# Check the overall sales calculation discrepancy
print("Recorded sales total:", round(df["sales_amount"].sum(), 2))
print("Expected sales total:", round(df["expected_sales"].sum(), 2))

print("\nSales difference statistics:")
print(df["sales_difference"].describe())

print("\nSample of 10 orders:")
print(
    df[
        ["quantity", "unit_price", "discount_pct",
         "expected_sales", "sales_amount", "sales_difference"]
    ].sample(10, random_state=42).round(2)
)
```

```python
# Remove temporary investigation columns
df = df.drop(
    columns=["expected_sales", "sales_difference"],
    errors="ignore"
)

print("Temporary columns removed.")
print("\nCurrent dataset shape:", df.shape)
print("\nCurrent columns:")
print(df.columns.tolist())
```

```python
# Inspect all column data types
print("Current data types:")
print(df.dtypes)

print("\nMissing values by column:")
print(df.isna().sum())
```

```python
# Convert whole-number columns to appropriate integer types
df["age"] = df["age"].astype("int64")
df["quantity"] = df["quantity"].astype("int64")

# Use nullable integer because shipping times contain missing values
df["days_to_ship"] = df["days_to_ship"].astype("Int64")

# Verify
print("Updated data types:")
print(df[["age", "quantity", "days_to_ship"]].dtypes)

print("\nMissing shipping times:",
      df["days_to_ship"].isna().sum())
```
